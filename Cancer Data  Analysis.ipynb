# INFRENTIAL AND PREDICTIVE ANALYSIS

```python
from scipy.stats import linregress
risk_factors=['Genetic_Risk','Air_Pollution', 'Alcohol_Use', 'Smoking', 'Obesity_Level']
title=['Genetic Risk','Air Pollution', 'Use of Alcohol', 'Smoking', 'Obesity Level']
colors=["Cyan","Yellow","Pink","Violet","Purple"]
plt.figure(figsize=(20,12))
for i,(factor,title,color) in enumerate(zip(risk_factors,title,colors),1):
    plt.subplot(2,3,i)
    x=df[factor] #input features
    y=df['Target_Severity_Score']
    slope,intercept,r_value,p_value,std_error=linregress(x,y)
    r_square=r_value**2
    sns.lineplot(x=factor,y="Target_Severity_Score",data=df,color=color)
    plt.plot(x,x*slope+intercept,color="Red",linewidth=2,label='Regression Line') #plotting regression line by formula y=mx+c
    plt.title(f"{title} vs Severity Score\n,R2-{r_square},slope-{slope}")
    plt.xlabel(factor)
    plt.ylabel("Target_Severity_Score")
    plt.legend()
plt.tight_layout()
plt.show()
```

## HYPERPARAMETER TUNITNG-GRID SEARCH CV

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import r2_score

# Define hyperparameter combinations
param_grid = {
    "n_estimators": [100, 200, 300],
    "max_depth": [5, 10, 15, 20, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4]
}

# Hyperparameter tuning
search = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=27),
    param_distributions=param_grid,
    n_iter=20,
    cv=5,
    scoring="r2",
    random_state=27,
    n_jobs=-1,
    verbose=1
)

# Train using your existing variables
search.fit(x_train, y_train)

# Best hyperparameters
print("Best Parameters:", search.best_params_)
print("Best CV R²:", search.best_score_)

# Best tuned model
best_model = search.best_estimator_

# Predictions
y_train_pred = best_model.predict(x_train)
y_test_pred = best_model.predict(x_test)

# Evaluation
train_r2 = r2_score(y_train, y_train_pred)
test_r2 = r2_score(y_test, y_test_pred)

print("Train R²:", round(train_r2, 4))
print("Test R²:", round(test_r2, 4))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score

# Prepare features and target
X = df1.drop(columns=[
    "Patient_ID", "Survival_Years",
    "Target_Severity_Score", "Treatment_Cost_USD"
])

y = df1["Survival_Years"]

# Split the data
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=27
)
# Encode categorical columns
categorical_cols = [
    "Gender", "Country_Region",
    "Cancer_Type", "Cancer_Stage"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"),
         categorical_cols)
    ],
    remainder="passthrough"
)
# Train Random Forest
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=300,
        max_depth=15,
        min_samples_split=10,
        min_samples_leaf=4,
        random_state=27,
        n_jobs=-1
    ))
])

model.fit(x_train, y_train)
# Evaluate the model
train_r2 = r2_score(
    y_train, model.predict(x_train)
)

test_r2 = r2_score(
    y_test, model.predict(x_test)
)

print("Train R²:", round(train_r2, 4))
print("Test R²:", round(test_r2, 4))
```

### Random Forest Regression — Interpretation

1. Weak feature–target relationships: Near-zero Pearson and Spearman correlations suggest limited predictive information about survival years.
2. Overfitting: Train R² = 0.2696 and test R² = -0.0051 indicate poor generalization to unseen data.
3. Missing or noisy information: Important clinical predictors may be absent, or survival years may have little systematic relationship with the available features.

Conclusion: The current Random Forest model cannot reliably predict survival years. Further investigation of the dataset and target variable is necessary before additional hyperparameter tuning.

# Testing overfitting by cross validation

```python
from sklearn.model_selection import cross_validate
import numpy as np

cv_results = cross_validate(
    model,
    x_train,
    y_train,
    cv=5,
    scoring="r2",
    return_train_score=True,
    n_jobs=1
)

train_scores = cv_results["train_score"]
val_scores = cv_results["test_score"]

print("Training R²:", np.round(train_scores, 4))
print("Validation R²:", np.round(val_scores, 4))

print("Mean Train R²:", round(train_scores.mean(), 4))
print("Mean Validation R²:", round(val_scores.mean(), 4))

print("Mean Gap:",
      round(train_scores.mean() - val_scores.mean(), 4))
```

### Random Forest – Cross-Validation Interpretation

* Mean Training R² = 0.2913: The model explains approximately 29.13% of the variation in training data.

* Mean Validation R² = -0.0048: The model performs slightly worse than simply predicting the mean on unseen data.

* Mean Gap = 0.2961: The substantial gap indicates overfitting.

* Consistent Results: All five validation scores are negative, confirming consistently poor generalization across the folds.

* Conclusion: The model shows some overfitting and weak overall predictive performance. Missing important predictive features may be contributing to the low R², but further investigation is needed.

# Assess whether higher treatment cost is associated with longer survival

```python
from scipy.stats import pearsonr,spearmanr
x=df["Treatment_Cost_USD"]
y=df["Survival_Years"]
#Performing pearson correlation test:
pearson_corr,pearson_p=pearsonr(x,y)
print(f"Pearson correlation coefficient: {pearson_corr}")
print(f"Pearson P-value: {pearson_p}")

#Performing Spearman correlation test:
spearman_corr,spearman_p=spearmanr(x,y)
print(f"Spearman correlation coefficient: {spearman_corr}")
print(f"Pearson P-value: {spearman_p}")

alpha=0.05
def interpret_corr(corr,p,method):
    if p<alpha: 
        print(f"{method},We reject the null hypothesis")
    else:
        print(f"{method},We failed to reject null hypothesis")
interpret_corr(pearson_corr,pearson_p,"Pearson")
interpret_corr(spearman_corr,spearman_p,"Spearman")
```

# Examine wheather higher genetic risk amplifies the negative effects of smoking on cancer severity

```python
df4=df.copy()
```

```python
import statsmodels.formula.api as smf
model=smf.ols("Target_Severity_Score ~ Genetic_Risk*Smoking	",data =df4).fit()
model.summary2()
```

```python
model.summary2().tables[1].loc["Genetic_Risk:Smoking"]
```

## Inference: Effect of Genetic Risk and Smoking on Cancer Severity

1. Genetic Risk: Positive coefficient (0.2009) with p < 0.001 indicates a statistically significant positive association with cancer severity.

2. Smoking: Positive coefficient (0.2036) with p < 0.001 indicates a statistically significant positive association with cancer severity.

3. Interaction Effect: The Genetic Risk × Smoking interaction has a coefficient of approximately -0.0002 and p = 0.6283 (> 0.05), indicating no statistically significant interaction between the two variables.

4. Model Performance: R² = 0.466, indicating that the model explains approximately 46.6% of the variation in cancer severity.

### Conclusion
Genetic risk and smoking are individually associated with higher cancer severity. However, there is no statistically significant evidence that their combined effect differs from the sum of their individual effects.
